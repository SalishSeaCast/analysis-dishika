```python
import pandas as pd
import numpy as np
files_to_check = [
    "/results/forcing/atmospheric/GEM2.5/gemlam/gemlam_y2008m08d11.nc",
    "/results/forcing/atmospheric/GEM2.5/gemlam/gemlam_y2008m08d12.nc",
]

for file in files_to_check:
    ds_check = xr.open_dataset(file)
    times = pd.DatetimeIndex(ds_check["time_counter"].values)

    print("\nFile:", file)
    print("Dataset sizes:", dict(ds_check.sizes))
    print("Number of timestamps:", len(times))
    print("First timestamp:", times[0])
    print("Last timestamp:", times[-1])
    print("Dates inside file:", times.normalize().unique())
    print("Duplicate timestamps:", times.duplicated().sum())
    print("Finite tair values:", np.isfinite(ds_check["tair"].values).sum())

    ds_check.close()
```
