```python
import arrow
import copy
import datetime as dt

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from pathlib import Path
import xarray as xr
import glob
```

```python
badfiles = ['gemlam_y2008m07d16.nc',
'gemlam_y2008m07d17.nc',
'gemlam_y2008m07d18.nc',
'gemlam_y2008m07d19.nc',
'gemlam_y2008m07d20.nc',
'gemlam_y2008m07d21.nc',
'gemlam_y2008m07d22.nc',
'gemlam_y2008m07d23.nc',
'gemlam_y2008m08d10.nc',
'gemlam_y2008m08d11.nc',
'gemlam_y2008m08d12.nc']
```

```python
mypath = Path('/results/forcing/atmospheric/GEM2.5/gemlam')
result_directory = Path('/ocean/dtaneja/MOAD/analysis-dishika/notebooks/TimeFixed')
for file in badfiles:
    year = int(file[8:12])
    month = int(file[13:15])
    day = int(file[16:18])

    input_file = mypath / file
    output_file = result_directory / file

    orig = xr.open_dataset(input_file)
    start = dt.datetime(year, month, day, 0)
    endtime = dt.datetime(year, month, day, 23)
    timerange = arrow.Arrow.range("hour",start,endtime,)

    newtimes = []
    for r in timerange:
        newtimes.append(r.naive)

    ds_new = orig.assign_coords(time_counter=newtimes)
    ds_new.attrs["history"] = (ds_new.attrs.get("history", "")+ f"\n{arrow.now()} Corrected time stamps")
    ds_new.to_netcdf(output_file)
```
